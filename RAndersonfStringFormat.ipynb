# Practice with Python f-String

### 1. Binary, Octal, and Hexadecimal
You can display numbers in different bases using `b` for binary, `o` for octal, or `x` for hexadecimal. You can use: print(f"Binary: {variableName:b}") to display the number in binary.

```python
# Type your code here
num = 42

print(f"Binary: {num:b}")
print(f"Octal: {num:o}")
print(f"Hexadecimal: {num:x}")
```

### 2. Width, Decimal Precision, Large numbers, Percentages

f-string syntax: f"text:format"

You can set up the format as:
* width size use n where n is an integer number
* decimal precision use .n where n is an integer number
* Large number use ,
* Percentage use %

TODO:
* Declare a variable distance to the sun as 93 millions (93,000,000), then print the variable in large number format.
* Declare variables cost = 1500 and revenue = 2500 and ratio = subtract cost from revenue, then divided by cost.
* print variable ratio as 2 decimal precision in 10 width size, left aligned.
* Print variable ratio as 1 decimal precision in a 10 width size, center aligned with percentage.

```python
# Write your code here
distanceToSun = 93000000
print(f"The distance to the sun: {distanceToSun:,}")

cost = 1500
revenue = 2500

ratio = (revenue - cost) / cost

print(f"Ratio: {ratio:<10.2f}")
print(f"Ratio: {ratio:^10.1%}")
```

## 3. Number Grouping with Underscores
Underscores can be used instead of commas for large numbers (e.g., for code readability), for example instead of 8,232,123 it will display it as 8_232_123. Follow instruction in the next cell.

```python
population = 83200000
# print population in large number comma format.

# print population in large number underscore format.
population = 83200000

print(f"Population With Commas: {population:,}")
print(f"Population With Underscores: {population:_}")
```
